```python
# =============================
# SPACESHIP TITANIC COMPARISON
# LightGBM vs TF-DF
# =============================
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.preprocessing import LabelEncoder
import lightgbm as lgb
import tensorflow_decision_forests as tfdf


train = pd.read_csv('/kaggle/input/spaceship-titanic/train.csv')
test = pd.read_csv('/kaggle/input/spaceship-titanic/test.csv')
submission = pd.read_csv('/kaggle/input/spaceship-titanic/sample_submission.csv')


def eda_report(train, cat_cols=None, num_cols=None, top_n=10):
    """
    Advanced EDA for categorical + numerical columns
    Args:
        df (pd.DataFrame): Dataframe
        cat_cols (list): List of categorical column names (optional)
        num_cols (list): List of numerical column names (optional)
        top_n (int): Top categories to show for categorical plots
    """

    # ==============================
    # Handle categorical variables
    # ==============================
    if cat_cols:
        print("\n" + "="*80)
        print("📊 CATEGORICAL FEATURES ANALYSIS")
        print("="*80)

        for col in cat_cols:
            print("\n" + "-"*60)
            print(f"🔹 Column: {col}")
            print("-"*60)

            # Missing & unique
            missing = train[col].isnull().sum()
            unique_vals = train[col].nunique()
            print(f"Missing values: {missing} ({missing/len(train)*100:.2f}%)")
            print(f"Unique categories: {unique_vals}")

            # Frequency table
            print("\nTop Categories:")
            print(train[col].value_counts().head(top_n))

            # Plot
            plt.figure(figsize=(10, 4))
            sns.countplot(
                data=train, 
                x=col, 
                order=train[col].value_counts().head(top_n).index,
                palette="viridis"
            )
            plt.title(f"Top {top_n} categories of {col}")
            plt.xticks(rotation=45, ha="right")
            plt.show()

    # ==============================
    # Handle numerical variables
    # ==============================
    if num_cols:
        print("\n" + "="*80)
        print("📈 NUMERICAL FEATURES ANALYSIS")
        print("="*80)

        for col in num_cols:
            print("\n" + "-"*60)
            print(f"🔹 Column: {col}")
            print("-"*60)

            # Missing & descriptive stats
            missing = train[col].isnull().sum()
            print(f"Missing values: {missing} ({missing/len(train)*100:.2f}%)")
            print("\nDescriptive Stats:")
            print(train[col].describe(percentiles=[0.25, 0.5, 0.75]))

            # Histogram + KDE
            plt.figure(figsize=(12, 5))

            plt.subplot(1, 2, 1)
            sns.histplot(train[col].dropna(), bins=30, kde=True, color="blue")
            plt.title(f"Distribution of {col}")

            # Boxplot
            plt.subplot(1, 2, 2)
            sns.boxplot(x=train[col], color="orange")
            plt.title(f"Boxplot of {col}")

            plt.show()

# ============================
# USAGE
# ============================

# Detect columns automatically if not provided
categorical_cols = train.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols   = train.select_dtypes(include=['int64', 'float64']).columns.tolist()

# Run full EDA
eda_report(train, cat_cols=categorical_cols, num_cols=numerical_cols, top_n=10)

# =============================
# 2. FEATURE ENGINEERING
# =============================
def feature_engineering(df):
    df[['CabinDeck','CabinNum','CabinSide']] = df['Cabin'].str.split('/', expand=True)
    spending_cols = ["RoomService","FoodCourt","ShoppingMall","Spa","VRDeck"]
    df["TotalSpending"] = df[spending_cols].sum(axis=1)
    df["Group"] = df["PassengerId"].apply(lambda x: x.split("_")[0])
    df["GroupSize"] = df.groupby("Group")["PassengerId"].transform("count")
    df["CabinNum"] = pd.to_numeric(df["CabinNum"], errors="coerce")
    return df

train = feature_engineering(train)
test = feature_engineering(test)

# =============================
# 3. HANDLE MISSING VALUES
# =============================
spending_cols = ["RoomService","FoodCourt","ShoppingMall","Spa","VRDeck","Age","TotalSpending","CabinNum"]
for col in spending_cols:
    train[col] = train[col].fillna(train[col].median())
    test[col] = test[col].fillna(train[col].median())

for col in ["HomePlanet","CryoSleep","Destination","VIP","CabinDeck","CabinSide"]:
    train[col] = train[col].fillna(train[col].mode()[0])
    test[col] = test[col].fillna(train[col].mode()[0])

# =============================
# 4. SPLIT DATA
# =============================
y = train["Transported"].astype(int)
X = train.drop(columns=["PassengerId","Cabin","Name","Transported","Group"])

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# ✅ Convert booleans to string for TF-DF
for col in ["CryoSleep", "VIP"]:
    if col in X_train.columns:
        X_train[col] = X_train[col].astype(str)
        X_val[col] = X_val[col].astype(str)
        test[col] = test[col].astype(str)

# =============================
# 5A. LIGHTGBM
# =============================
X_train_lgb = X_train.copy()
X_val_lgb = X_val.copy()
categorical_cols = ["HomePlanet","CryoSleep","Destination","VIP","CabinDeck","CabinSide"]

# ✅ FIXED: use separate encoders and re-use on test
encoders = {}
for col in categorical_cols:
    le = LabelEncoder()
    X_train_lgb[col] = le.fit_transform(X_train_lgb[col].astype(str))
    X_val_lgb[col] = le.transform(X_val_lgb[col].astype(str))
    encoders[col] = le  # save encoder for later

model_lgb = lgb.LGBMClassifier(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=-1,
    colsample_bytree=0.8,
    subsample=0.8,
    random_state=42
)
model_lgb.fit(X_train_lgb, y_train)

preds_lgb = model_lgb.predict(X_val_lgb)
proba_lgb = model_lgb.predict_proba(X_val_lgb)[:,1]

acc_lgb = accuracy_score(y_val, preds_lgb)
auc_lgb = roc_auc_score(y_val, proba_lgb)

print("LightGBM -> Accuracy:", acc_lgb, "ROC-AUC:", auc_lgb)

# =============================
# 5B. TF-DF
# =============================
train_tfdf = X_train.copy()
train_tfdf["Transported"] = y_train
val_tfdf = X_val.copy()
val_tfdf["Transported"] = y_val

train_ds = tfdf.keras.pd_dataframe_to_tf_dataset(train_tfdf, label="Transported")
val_ds = tfdf.keras.pd_dataframe_to_tf_dataset(val_tfdf, label="Transported")

model_tfdf = tfdf.keras.GradientBoostedTreesModel(
    num_trees=500,
    max_depth=8,
    subsample=0.8,
    random_seed=42
)
model_tfdf.fit(train_ds)

val_preds = model_tfdf.predict(val_ds)
val_preds = np.argmax(val_preds, axis=1)

acc_tfdf = accuracy_score(y_val, val_preds)
auc_tfdf = roc_auc_score(y_val, val_preds)

print("TF-DF -> Accuracy:", acc_tfdf, "ROC-AUC:", auc_tfdf)

# =============================
# 6. CHOOSE BEST MODEL & SUBMIT
# =============================
drop_cols = ["PassengerId","Cabin","Name","Group"]  # ✅ same as training

if auc_lgb > auc_tfdf:
    print("🔥 LightGBM performs better on validation")
    best_model = "LGBM"

    # ✅ FIXED: drop extra cols in test
    test_lgb = test.drop(columns=drop_cols, errors="ignore").copy()

    # ✅ apply stored encoders
    for col in categorical_cols:
        test_lgb[col] = encoders[col].transform(test_lgb[col].astype(str))

    test_preds = model_lgb.predict(test_lgb)

else:
    print("🚀 TF-DF performs better on validation")
    best_model = "TF-DF"

    test_tfdf = test.drop(columns=drop_cols, errors="ignore").copy()
    test_ds = tfdf.keras.pd_dataframe_to_tf_dataset(test_tfdf)
    test_preds = model_tfdf.predict(test_ds)
    test_preds = np.argmax(test_preds, axis=1)

submission["Transported"] = test_preds.astype(bool)
submission.to_csv("submission_best.csv", index=False)

print(f"✅ Submission file created with {best_model}: submission_best.csv")
```
