### EDA Stress Level Dataset

```python
# =============================
# eda Stress Level Dataset
# =============================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis

# ======================
# STEP 1: LOAD DATA
# ======================
df = pd.read_csv("/kaggle/input/student-stress-monitoring-datasets/StressLevelDataset.csv") 

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:\n", df.head())
print("\nData Info:")
print(df.info())

# ======================
# STEP 2: MISSING VALUES
# ======================
print("\nMissing values per column:")
print(df.isnull().sum())

plt.figure(figsize=(10, 5))
sns.heatmap(df.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Values Heatmap")
plt.show()

# ======================
# STEP 3: SUMMARY STATISTICS
# ======================
print("\nSummary Statistics for Numerical Features:")
print(df.describe().T)

# ======================
# STEP 4: TARGET DISTRIBUTION
# ======================
if "Stress Level" in df.columns or "stress" in df.columns or "label" in df.columns:
    target_col = "Stress Level" if "Stress Level" in df.columns else "stress"
    plt.figure(figsize=(6, 4))
    sns.countplot(x=df[target_col], palette="Set2")
    plt.title("Target Distribution")
    plt.show()
    print("\nTarget Value Counts:")
    print(df[target_col].value_counts())

# ======================
# STEP 5: NUMERICAL DISTRIBUTIONS
# ======================
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col].dropna(), kde=True, bins=30, color="skyblue")
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
    print(f"{col} → Skewness: {skew(df[col].dropna()):.2f}, Kurtosis: {kurtosis(df[col].dropna()):.2f}")

# ======================
# STEP 6: CATEGORICAL DISTRIBUTIONS
# ======================
cat_cols = df.select_dtypes(include="object").columns

for col in cat_cols:
    plt.figure(figsize=(8, 4))
    sns.countplot(y=col, data=df, order=df[col].value_counts().index[:10], palette="coolwarm")
    plt.title(f"Top Categories of {col}")
    plt.show()
    print(f"{col} → Top 5 categories:\n{df[col].value_counts().head()}")

# ======================
# STEP 7: CORRELATION ANALYSIS
# ======================
if len(num_cols) > 1:
    corr = df[num_cols].corr()
    plt.figure(figsize=(10, 6))
    sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
    plt.title("Correlation Heatmap")
    plt.show()

# ======================
# STEP 8: FEATURE vs TARGET RELATIONSHIPS
# ======================
if "Stress Level" in df.columns:
    target = "Stress Level"
    
    for col in num_cols:
        if col != target:
            plt.figure(figsize=(8, 4))
            sns.boxplot(x=df[target], y=df[col], palette="Set3")
            plt.title(f"{col} vs {target}")
            plt.show()

    for col in cat_cols:
        plt.figure(figsize=(8, 4))
        sns.countplot(x=df[col], hue=df[target], palette="viridis")
        plt.title(f"{col} vs {target}")
        plt.xticks(rotation=30)
        plt.show()

# ======================
# STEP 9: OUTLIER DETECTION
# ======================
for col in num_cols:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[col], color="lightcoral")
    plt.title(f"Outliers in {col}")
    plt.show()

# ======================
# STEP 10: INSIGHTS (Auto Summary)
# ======================
print("\nKEY INSIGHTS:")
if "Stress Level" in df.columns:
    print("- Stress levels are imbalanced across classes → consider SMOTE or class weighting in ML.")
if len(num_cols) > 0:
    print("- Some numerical features show high skew → log-transform may help.")
if df.isnull().sum().sum() > 0:
    print("- Missing values exist → need imputation before ML.")
if len(cat_cols) > 0:
    print("- Many categorical features → encoding strategies (Label Encoding, One-Hot, Target Encoding).")
print("- Outliers detected in numerical features → could use RobustScaler or trimming.")
```

```python
# =============================
# STRESS LEVEL PREDICTION PIPELINE
# =============================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

import lightgbm as lgb
import xgboost as xgb
from catboost import CatBoostClassifier

# ======================
# STEP 1: LOAD DATA
# ======================
df = pd.read_csv("/kaggle/input/student-stress-monitoring-datasets/StressLevelDataset.csv")  # change if needed

print("Dataset shape:", df.shape)
print("Columns:", df.columns)

# Identify target column
target = "stress_level" if "stress_level" in df.columns else "stress"

# ======================
# STEP 2: PREPROCESSING
# ======================

# Handle missing values
df = df.fillna(-999)

# Separate features & target
X = df.drop(columns=[target, "id"], errors="ignore")
y = df[target]

# Encode categorical features
for col in X.select_dtypes(include="object").columns:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col].astype(str))

# Scale numerical features
scaler = StandardScaler()
X[X.columns] = scaler.fit_transform(X)

# ======================
# STEP 3: FEATURE ENGINEERING
# ======================

def add_features(df):
    df = df.copy()
    # Example: interactions
    if {"Sleep Hours", "Work Hours"}.issubset(df.columns):
        df["work_sleep_ratio"] = df["Work Hours"] / (df["Sleep Hours"] + 1e-3)
    if {"Heart Rate", "Breathing Rate"}.issubset(df.columns):
        df["heart_breath_ratio"] = df["Heart Rate"] / (df["Breathing Rate"] + 1e-3)
    if "Sleep Hours" in df.columns:
        df["sleep_quality"] = np.log1p(df["Sleep Hours"])
    return df

X = add_features(X)

# ======================
# STEP 4: TRAIN/VAL SPLIT
# ======================
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# ======================
# STEP 5: MODELS
# ======================

# ---- LightGBM ----
lgb_model = lgb.LGBMClassifier(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=-1,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)
lgb_model.fit(X_train, y_train, eval_set=[(X_val, y_val)], eval_metric="logloss",
              callbacks=[lgb.early_stopping(100)])

# ---- XGBoost ----
xgb_model = xgb.XGBClassifier(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="mlogloss",
    random_state=42,
    tree_method="hist"
)
xgb_model.fit(X_train, y_train, eval_set=[(X_val, y_val)], early_stopping_rounds=100, verbose=200)

# ---- CatBoost ----
cat_model = CatBoostClassifier(
    iterations=1000,
    learning_rate=0.05,
    depth=6,
    eval_metric="Accuracy",
    random_seed=42,
    verbose=200,
    early_stopping_rounds=100
)
cat_model.fit(X_train, y_train, eval_set=(X_val, y_val))

# ======================
# STEP 6: EVALUATION
# ======================

models = {"LightGBM": lgb_model, "XGBoost": xgb_model, "CatBoost": cat_model}

for name, model in models.items():
    preds = model.predict(X_val)
    print(f"\n{name} Performance:")
    print("Accuracy:", accuracy_score(y_val, preds))
    print("F1 Score:", f1_score(y_val, preds, average="weighted"))
    print(classification_report(y_val, preds))

    # Confusion Matrix
    cm = confusion_matrix(y_val, preds)
    plt.figure(figsize=(5, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.show()

# ======================
# STEP 7: FEATURE IMPORTANCE
# ======================
lgb_imp = pd.Series(lgb_model.feature_importances_, index=X.columns).sort_values(ascending=False)
plt.figure(figsize=(10, 6))
sns.barplot(x=lgb_imp[:15], y=lgb_imp.index[:15])
plt.title("Top 15 Feature Importances (LightGBM)")
plt.show()
```
