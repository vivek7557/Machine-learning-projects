# EDA for asia-cup-cricket-1984-to-2022

This notebook includes:
- Dataset overview and summary statistics
- Missing value analysis
- Distribution plots for numeric columns
- Boxplots for outliers
- Correlation heatmap
- Countplots for categorical columns
- Pairplot for top numeric columns
- Grouped boxplots (numeric vs categorical)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

df = pd.read_csv("/kaggle/input/asia-cup-cricket-1984-to-2022/asiacup.csv")
print("Shape:", df.shape)
print("\nDtypes:\n", df.dtypes)
df.head()
```

```python
plt.figure(figsize=(10,6))
plt.matshow(df.isnull(), cmap="viridis", fignum=1)
plt.title("Missing Value Heatmap")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```

```python
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()

for col in num_cols:
    fig, ax = plt.subplots(1,2, figsize=(12,4))
    sns.histplot(df[col].dropna(), kde=True, bins=30, color="skyblue", ax=ax[0])
    ax[0].set_title(f"Distribution: {col}")
    sns.boxplot(x=df[col], color="orange", ax=ax[1])
    ax[1].set_title(f"Boxplot: {col}")
    plt.show()
```

```python
if len(num_cols) > 1:
    corr = df[num_cols].corr()
    plt.figure(figsize=(10,8))
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", square=True, cbar=True)
    plt.title("Correlation Heatmap")
    plt.show()
```

```python
cat_cols = df.select_dtypes(include=['object','category']).columns.tolist()

for col in cat_cols[:6]:
    plt.figure(figsize=(8,4))
    sns.countplot(y=df[col], order=df[col].value_counts().index, palette="viridis")
    plt.title(f"Countplot: {col}")
    plt.show()
```

```python
if len(num_cols) > 1:
    top_num = df[num_cols].var().sort_values(ascending=False).head(4).index.tolist()
    sns.pairplot(df[top_num].dropna(), diag_kind="kde", corner=True)
    plt.show()
```

```python
if cat_cols and num_cols:
    for num in num_cols[:2]:
        plt.figure(figsize=(10,5))
        sns.boxplot(x=df[cat_cols[0]], y=df[num], palette="Set3")
        plt.title(f"{num} grouped by {cat_cols[0]}")
        plt.xticks(rotation=45)
        plt.show()
```

```python
# cricket_enhanced_eda_kaggle.py
# Enhanced version with attractive visualizations

import os
import zipfile
import traceback
from pathlib import Path
from typing import Dict, List, Optional

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.patches import Rectangle
import matplotlib.patches as mpatches

# Set attractive styling
plt.style.use('default')
sns.set_palette("husl")
plt.rcParams.update({
    'font.size': 11,
    'axes.titlesize': 14,
    'axes.labelsize': 12,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'legend.fontsize': 10,
    'figure.titlesize': 16,
    'font.family': 'sans-serif'
})

# Cricket-themed color palette
CRICKET_COLORS = {
    'primary': '#1f77b4',      # Cricket blue
    'secondary': '#ff7f0e',    # Orange
    'success': '#2ca02c',      # Green
    'danger': '#d62728',       # Red
    'warning': '#ff7f0e',      # Orange
    'info': '#17a2b8',         # Light blue
    'dark': '#343a40',         # Dark gray
    'light': '#f8f9fa'         # Light gray
}

# --- Config ---
DATA_DIR = Path("/kaggle/input/asia-cup-cricket-1984-to-2022")
OUT_DIR = Path("./outputs")

COMMON_RUN_COLS = ["runs","runs_scored","total_runs","bat_runs"]
COMMON_WKT_COLS = ["wickets","wkts","wickets_taken","total_wkts"]
COMMON_CATCH_COLS = ["catches","ct","catch","catches_taken"]
COMMON_STUMP_COLS = ["stumpings","stumps"]
COMMON_FORMAT_COLS = ["format","match_type","type","class","matchformat"]
COMMON_PLAYER_COLS = ["player","player_name","name","full_name"]

MAX_CAT_VALUECOUNTS = 20
MAX_NUM_HIST = 6

# --- Helpers ---
def ensure_dir(p: Path):
    p.mkdir(parents=True, exist_ok=True)

def extract_all_zips(data_dir: Path):
    for z in data_dir.rglob("*.zip"):
        try:
            print(f"🔄 Extracting {z} ...")
            with zipfile.ZipFile(z, "r") as zip_ref:
                zip_ref.extractall(z.parent)
        except Exception as e:
            print(f"  ❌ Failed to extract {z}: {e}")

def find_csvs(data_dir: Path) -> List[Path]:
    return sorted([p for p in data_dir.rglob("*.csv")])

def try_read_csv(p: Path) -> pd.DataFrame:
    try:
        return pd.read_csv(p)
    except Exception:
        try:
            return pd.read_csv(p, encoding="latin-1")
        except Exception:
            return pd.read_csv(p, encoding="latin-1", low_memory=False)

def find_best_column(df: pd.DataFrame, candidates: List[str]) -> Optional[str]:
    low = {c.lower(): c for c in df.columns}
    for cand in candidates:
        if cand.lower() in low:
            return low[cand.lower()]
    for cand in candidates:
        for col_low, col_orig in low.items():
            if cand.lower() in col_low:
                return col_orig
    return None

def to_numeric_safe(s: pd.Series) -> pd.Series:
    return pd.to_numeric(s, errors="coerce")

def sanitize_filename(s: str) -> str:
    return "".join(c if c.isalnum() or c in "-_." else "_" for c in s)[:200]

# --- Enhanced Top-10 logic ---
def compute_top10s(df: pd.DataFrame) -> Dict[str, Dict[str, Optional[pd.DataFrame]]]:
    res = {}
    player_col = find_best_column(df, COMMON_PLAYER_COLS) or df.columns[0]
    format_col = find_best_column(df, COMMON_FORMAT_COLS)
    runs_col = find_best_column(df, COMMON_RUN_COLS)
    wickets_col = find_best_column(df, COMMON_WKT_COLS)
    catches_col = find_best_column(df, COMMON_CATCH_COLS)
    stump_col = find_best_column(df, COMMON_STUMP_COLS)

    df2 = df.copy()
    df2["_catches_"] = to_numeric_safe(df2[catches_col]).fillna(0) if catches_col else 0
    df2["_stumps_"] = to_numeric_safe(df2[stump_col]).fillna(0) if stump_col else 0
    df2["_dismissals_"] = df2[["_catches_","_stumps_"]].sum(axis=1)

    def top_by(column: str, fmt: Optional[str] = None):
        if column is None or column not in df2.columns:
            return None
        s = df2.copy()
        if fmt and format_col:
            s = s[s[format_col].astype(str).str.lower() == fmt.lower()]
        s["_stat"] = to_numeric_safe(s[column])
        s = s.dropna(subset=["_stat"])
        if s.empty:
            return None
        out = s.groupby(player_col)["_stat"].sum().reset_index()
        out = out.sort_values("_stat", ascending=False).head(10)
        return out.rename(columns={player_col:"player","_stat":"stat"}).reset_index(drop=True)

    for fmt in ["ODI","T20"]:
        res[fmt] = {
            "batsmen": top_by(runs_col, fmt),
            "bowlers": top_by(wickets_col, fmt),
            "keepers": top_by("_dismissals_", fmt)
        }
    res["global"] = {
        "batsmen": top_by(runs_col, None),
        "bowlers": top_by(wickets_col, None),
        "keepers": top_by("_dismissals_", None)
    }
    return res

# --- Enhanced Plotting Functions ---
def create_gradient_background(ax, color_start='#f0f8ff', color_end='#e6f3ff'):
    """Add gradient background to plot"""
    gradient = np.linspace(0, 1, 256).reshape(1, -1)
    gradient = np.vstack((gradient, gradient))
    ax.imshow(gradient, aspect='auto', extent=[ax.get_xlim()[0], ax.get_xlim()[1], 
                                             ax.get_ylim()[0], ax.get_ylim()[1]], 
              alpha=0.3, cmap=plt.cm.Blues)

def add_cricket_watermark(ax):
    """Add subtle cricket-themed watermark"""
    ax.text(0.95, 0.95, '🏏', transform=ax.transAxes, fontsize=20, 
            alpha=0.1, ha='right', va='top')

def plot_save_hist(series: pd.Series, title: str, outpath: Path, bins: int = 30):
    """Enhanced histogram with attractive styling"""
    fig, ax = plt.subplots(figsize=(10, 6))
    
    # Create histogram with gradient colors
    n, bins_edges, patches = ax.hist(series.dropna(), bins=bins, alpha=0.8, 
                                    edgecolor='white', linewidth=1.2)
    
    # Color gradient for bars
    colors = plt.cm.viridis(np.linspace(0.3, 0.9, len(patches)))
    for patch, color in zip(patches, colors):
        patch.set_facecolor(color)
    
    # Styling
    ax.set_title(title, fontsize=16, fontweight='bold', pad=20, 
                color=CRICKET_COLORS['dark'])
    ax.set_xlabel(series.name or 'Values', fontsize=12, fontweight='medium')
    ax.set_ylabel('Frequency', fontsize=12, fontweight='medium')
    
    # Grid and spines
    ax.grid(True, alpha=0.3, linestyle='--')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_color(CRICKET_COLORS['dark'])
    ax.spines['bottom'].set_color(CRICKET_COLORS['dark'])
    
    # Add stats box
    stats_text = f'Count: {len(series.dropna())}\nMean: {series.mean():.2f}\nStd: {series.std():.2f}'
    ax.text(0.75, 0.75, stats_text, transform=ax.transAxes, 
            bbox=dict(boxstyle='round,pad=0.5', facecolor='white', alpha=0.8),
            fontsize=10, verticalalignment='top')
    
    add_cricket_watermark(ax)
    plt.tight_layout()
    plt.savefig(outpath, dpi=300, bbox_inches='tight', facecolor='white')
    plt.show()
    plt.close()

def plot_corr_heatmap(corr: pd.DataFrame, outpath: Path):
    """Enhanced correlation heatmap with attractive styling"""
    fig, ax = plt.subplots(figsize=(max(8, corr.shape[1]*0.8), max(6, corr.shape[0]*0.6)))
    
    # Create mask for upper triangle
    mask = np.triu(np.ones_like(corr, dtype=bool))
    
    # Enhanced heatmap
    sns.heatmap(corr, mask=mask, annot=True, cmap='RdYlBu_r', center=0,
                square=True, linewidths=0.5, cbar_kws={"shrink": 0.8},
                fmt='.2f', ax=ax)
    
    ax.set_title('Correlation Matrix', fontsize=18, fontweight='bold', 
                pad=20, color=CRICKET_COLORS['dark'])
    
    # Rotate labels for better readability
    plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
    plt.setp(ax.get_yticklabels(), rotation=0)
    
    add_cricket_watermark(ax)
    plt.tight_layout()
    plt.savefig(outpath, dpi=300, bbox_inches='tight', facecolor='white')
    plt.show()
    plt.close()

def plot_save_bar(df: pd.DataFrame, x_col: str, y_col: str, title: str, outpath: Path):
    """Enhanced bar chart with cricket-themed styling"""
    fig, ax = plt.subplots(figsize=(12, 8))
    
    # Create bars with gradient colors
    bars = ax.bar(range(len(df)), df[y_col], 
                  color=plt.cm.Set3(np.linspace(0, 1, len(df))),
                  edgecolor='white', linewidth=1.5, alpha=0.9)
    
    # Add value labels on bars
    for i, (bar, value) in enumerate(zip(bars, df[y_col])):
        height = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2., height + height*0.01,
                f'{value:.0f}', ha='center', va='bottom', fontweight='bold', fontsize=10)
    
    # Styling
    ax.set_title(title, fontsize=18, fontweight='bold', pad=25, 
                color=CRICKET_COLORS['dark'])
    ax.set_xlabel('Players', fontsize=14, fontweight='medium', labelpad=15)
    ax.set_ylabel('Statistics', fontsize=14, fontweight='medium', labelpad=15)
    
    # Set x-axis labels
    ax.set_xticks(range(len(df)))
    ax.set_xticklabels(df[x_col], rotation=45, ha='right', fontsize=11)
    
    # Grid and styling
    ax.grid(True, alpha=0.3, linestyle='--', axis='y')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_color(CRICKET_COLORS['dark'])
    ax.spines['bottom'].set_color(CRICKET_COLORS['dark'])
    
    # Add ranking numbers
    for i, bar in enumerate(bars):
        ax.text(bar.get_x() + bar.get_width()/2., bar.get_height()/2,
                f'#{i+1}', ha='center', va='center', fontweight='bold', 
                fontsize=12, color='white', 
                bbox=dict(boxstyle='circle,pad=0.3', facecolor=CRICKET_COLORS['dark'], alpha=0.8))
    
    add_cricket_watermark(ax)
    plt.tight_layout()
    plt.savefig(outpath, dpi=300, bbox_inches='tight', facecolor='white')
    plt.show()
    plt.close()

def create_data_overview_dashboard(df: pd.DataFrame, outpath: Path):
    """Create an attractive dashboard showing data overview"""
    fig = plt.figure(figsize=(16, 12))
    
    # Create subplots
    gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)
    
    # 1. Dataset Info
    ax1 = fig.add_subplot(gs[0, 0])
    info_text = f"""Dataset Overview
    
Rows: {df.shape[0]:,}
Columns: {df.shape[1]:,}
Memory: {df.memory_usage(deep=True).sum() / 1024**2:.1f} MB
Missing Values: {df.isnull().sum().sum():,}
Duplicate Rows: {df.duplicated().sum():,}"""
    
    ax1.text(0.05, 0.95, info_text, transform=ax1.transAxes, fontsize=12,
             verticalalignment='top', fontfamily='monospace',
             bbox=dict(boxstyle='round,pad=1', facecolor=CRICKET_COLORS['light'], alpha=0.8))
    ax1.set_xlim(0, 1)
    ax1.set_ylim(0, 1)
    ax1.axis('off')
    ax1.set_title('📊 Dataset Summary', fontsize=14, fontweight='bold', pad=20)
    
    # 2. Data types distribution
    ax2 = fig.add_subplot(gs[0, 1])
    dtype_counts = df.dtypes.value_counts()
    colors = plt.cm.Set3(np.linspace(0, 1, len(dtype_counts)))
    wedges, texts, autotexts = ax2.pie(dtype_counts.values, labels=dtype_counts.index, 
                                      autopct='%1.1f%%', colors=colors, startangle=90)
    ax2.set_title('🔧 Data Types Distribution', fontsize=14, fontweight='bold', pad=20)
    
    # 3. Missing values heatmap
    ax3 = fig.add_subplot(gs[0, 2])
    missing_data = df.isnull().sum()
    if missing_data.sum() > 0:
        missing_top = missing_data[missing_data > 0].sort_values(ascending=False).head(10)
        bars = ax3.barh(range(len(missing_top)), missing_top.values, 
                       color=CRICKET_COLORS['danger'], alpha=0.7)
        ax3.set_yticks(range(len(missing_top)))
        ax3.set_yticklabels(missing_top.index, fontsize=10)
        ax3.set_xlabel('Missing Count')
        for i, v in enumerate(missing_top.values):
            ax3.text(v + max(missing_top.values)*0.01, i, str(v), 
                    va='center', fontweight='bold', fontsize=10)
    else:
        ax3.text(0.5, 0.5, '✅ No Missing Values!', transform=ax3.transAxes,
                ha='center', va='center', fontsize=16, fontweight='bold',
                color=CRICKET_COLORS['success'])
    ax3.set_title('🚫 Missing Values', fontsize=14, fontweight='bold', pad=20)
    
    # 4-6. Numeric columns distribution (if available)
    numeric_cols = df.select_dtypes(include=[np.number]).columns[:3]
    for i, col in enumerate(numeric_cols):
        ax = fig.add_subplot(gs[1, i])
        df[col].hist(ax=ax, bins=20, alpha=0.7, color=CRICKET_COLORS['primary'])
        ax.set_title(f'📈 {col} Distribution', fontsize=12, fontweight='bold')
        ax.grid(True, alpha=0.3)
    
    # 7-9. Categorical columns (if available)
    cat_cols = df.select_dtypes(include=['object']).columns[:3]
    for i, col in enumerate(cat_cols):
        ax = fig.add_subplot(gs[2, i])
        top_values = df[col].value_counts().head(8)
        bars = ax.bar(range(len(top_values)), top_values.values, 
                     color=plt.cm.viridis(np.linspace(0.3, 0.9, len(top_values))))
        ax.set_xticks(range(len(top_values)))
        ax.set_xticklabels(top_values.index, rotation=45, ha='right', fontsize=10)
        ax.set_title(f'🏷️ Top {col} Values', fontsize=12, fontweight='bold')
        ax.grid(True, alpha=0.3, axis='y')
    
    plt.suptitle('🏏 Cricket Data Analysis Dashboard', fontsize=20, fontweight='bold', y=0.95)
    plt.savefig(outpath, dpi=300, bbox_inches='tight', facecolor='white')
    plt.show()
    plt.close()

# --- Enhanced Analyzer ---
def analyze_csv(p: Path, out_dir: Path):
    print(f"\n🏏 --- Processing {p} ---")
    out_dir = out_dir / p.stem
    ensure_dir(out_dir)

    df = try_read_csv(p)
    print(f"   📝 Shape: {df.shape}")

    # Save basic outputs
    (out_dir / "preview.csv").write_text(df.head(50).to_csv(index=False))
    (out_dir / "columns.txt").write_text("\n".join([f"{c}: {str(t)}" for c, t in df.dtypes.items()]))
    (out_dir / "shape.txt").write_text(str(df.shape))
    df.isnull().sum().to_csv(out_dir / "missing_values.csv")
    df.describe(include="all").T.to_csv(out_dir / "describe.csv")

    # Create dashboard
    try:
        create_data_overview_dashboard(df, out_dir / "dashboard_overview.png")
        print(f"   ✅ Dashboard created")
    except Exception as e:
        print(f"   ❌ Dashboard failed: {e}")

    # Enhanced histograms
    num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    print(f"   📊 Creating {min(len(num_cols), MAX_NUM_HIST)} histograms...")
    for i, c in enumerate(num_cols[:MAX_NUM_HIST]):
        plot_save_hist(df[c], f"Distribution of {c}", 
                      out_dir / f"hist_{i+1}__{sanitize_filename(c)}.png")

    # Enhanced correlation analysis
    if len(num_cols) >= 2:
        print(f"   🔗 Creating correlation analysis...")
        corr = df[num_cols].corr()
        corr.to_csv(out_dir / "correlation_matrix.csv")
        plot_corr_heatmap(corr, out_dir / "correlation_heatmap.png")

    # Enhanced Top-10s with better visualizations
    try:
        print(f"   🏆 Computing top performers...")
        tops = compute_top10s(df)
        for fmt, mapping in tops.items():
            if mapping is None: 
                continue
            fmt_dir = out_dir / f"top10_{fmt.lower()}"
            ensure_dir(fmt_dir)
            for kind, tbl in mapping.items():
                if tbl is None: 
                    continue
                tbl.to_csv(fmt_dir / f"{kind}.csv", index=False)
                try:
                    plot_save_bar(tbl, "player", "stat", 
                                 f"🏏 Top-10 {kind.title()} - {fmt.upper()}", 
                                 fmt_dir / f"{kind}.png")
                except Exception as e:
                    print(f"     ❌ Failed barplot: {kind} {fmt} - {e}")
    except Exception as e:
        print(f"   ❌ Top-10 analysis failed: {e}")

    print(f"   ✅ Finished {p.name}, outputs saved to {out_dir}")

# --- Main Execution ---
def main():
    print("🏏" + "="*60)
    print("    ENHANCED CRICKET DATA ANALYSIS")
    print("="*63 + "🏏")
    
    ensure_dir(OUT_DIR)
    
    print(f"\n📁 Extracting data from: {DATA_DIR}")
    extract_all_zips(DATA_DIR)

    csvs = find_csvs(DATA_DIR)
    print(f"\n📋 Found {len(csvs)} CSV files:")
    for i, csv_file in enumerate(csvs, 1):
        print(f"    {i}. {csv_file.name}")

    print(f"\n🚀 Starting analysis...")
    for i, p in enumerate(csvs, 1):
        print(f"\n[{i}/{len(csvs)}] Processing: {p.name}")
        try:
            analyze_csv(p, OUT_DIR)
        except Exception as e:
            print(f"❌ Error processing {p}: {e}")
            traceback.print_exc()

    print(f"\n🎉 Analysis Complete!")
    print(f"📊 All outputs saved to: {OUT_DIR}")
    print("🏏" + "="*60)

if __name__ == "__main__":
    main()
```
